```python

import sys
from pathlib import Path
import matplotlib.pyplot as plt
from medmnist import PneumoniaMNIST

sys.path.insert(0, str(Path.cwd().resolve()))
from src.preprocessing import preprocess

Path("outputs/figures").mkdir(parents=True, exist_ok=True)

dataset = PneumoniaMNIST(
    split="train", download=True, size=28
)

image = dataset.imgs[0]
methods = ["original", "clahe", "hist_eq"]

fig, axes = plt.subplots(2, 3, figsize=(12, 7))

for col, method in enumerate(methods):
    processed = preprocess(image, method)

    axes[0, col].imshow(
        processed, cmap="gray", vmin=0, vmax=255
    )
    axes[0, col].set_title(method)
    axes[0, col].axis("off")

    axes[1, col].hist(
        processed.ravel(),
        bins=32,
        range=(0, 256)
    )
    axes[1, col].set_xlim(0, 255)
    axes[1, col].set_title(f"{method} histogram")

plt.tight_layout()
plt.savefig(
    "outputs/figures/preprocessing_comparison.png",
    dpi=200
)
plt.show()
```
