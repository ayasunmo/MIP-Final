```python

import numpy as np
import matplotlib.pyplot as plt
from medmnist import PneumoniaMNIST

datasets = {
    split: PneumoniaMNIST(
        split=split,
        download=True,
        size=28
    )
    for split in ["train", "val", "test"]
}

for split, ds in datasets.items():
    images = ds.imgs
    labels = ds.labels.flatten()
    counts = np.bincount(labels, minlength=2)

    print(f"\n{split.upper()}")
    print("Image shape:", images.shape)
    print("Image dtype:", images.dtype)
    print("Normal:", counts[0])
    print("Pneumonia:", counts[1])
    print("Class ratios:", counts / len(labels))

# Show 5 images from each class
train = datasets["train"]
labels = train.labels.flatten()

fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for row, class_id in enumerate([0, 1]):
    indices = np.flatnonzero(labels == class_id)[:5]

    for col, idx in enumerate(indices):
        axes[row, col].imshow(
            train.imgs[idx],
            cmap="gray",
            vmin=0,
            vmax=255
        )
        axes[row, col].set_title(
            "Normal" if class_id == 0 else "Pneumonia"
        )
        axes[row, col].axis("off")

plt.tight_layout()
plt.savefig("outputs/figures/sample_xrays.png", dpi=200)
plt.show()

# Pixel intensity distribution
plt.figure(figsize=(7, 4))
plt.hist(train.imgs.flatten(), bins=50)
plt.title("Training X-ray Pixel Intensities")
plt.xlabel("Pixel intensity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("outputs/figures/pixel_histogram.png", dpi=200)
plt.show()
```
